```python
import numpy as np
from scipy.integrate import solve_ivp

m = 500
g = 9.81 
D = 0.14

def equations(t, z):
    x, y, vx, vy = z
    v = np.hypot(vx,vy)
    ax = -(D/m)*v*vx
    ay = -g - (D/m) * v *vy
    return[vx,vy,ax,ay]
def ground_impact(t,z):
    return z[1] 

ground_impact.terminal = True
ground_impact.direction = -1

z0 = [0, 1000, 1000,0] 
sol = solve_ivp(
    equations, [0,100], z0,events = ground_impact, rtol = 1e-8
)
t_impact = sol.t_events[0][0]
x_impact = sol.y_events[0][0][0]
print (f"Time to impact: {t_impact:.2f} s")
print (f"Horizontal Distance: {x_impact:.2f} m")
```
